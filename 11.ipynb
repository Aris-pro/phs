```python
import numpy as np
import pandas as pd
df = pd.read_csv("lab1.csv")
print(df.head())
df = df.iloc[:100]
X = df.iloc[:, 0:2].values
y = df.iloc[:, 4].values
y = np.where(y == "setosa", 0, 1)

class Adaline:

    def __init__(self, learning_rate=0.01, n_iterations=100):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None

    def fit(self, X, y):

        self.weights = np.zeros(1 + X.shape[1])
        for _ in range(self.n_iterations):

            net_input = np.dot(X, self.weights[1:]) + self.weights[0]
            predicted = net_input
            errors = y - predicted
            self.weights[1:] += self.learning_rate * np.dot(X.T, errors)
            self.weights[0] += self.learning_rate * errors.sum()

    def predict(self, X):
        net_input = np.dot(X, self.weights[1:]) + self.weights[0]
        return np.where(net_input >= 0.5, 1, 0)

adaline = Adaline(learning_rate=0.0001, n_iterations=50)
adaline.fit(X, y)

print("Weights:", adaline.weights)
sample = np.array([5.1, 3.5])
prediction = adaline.predict(sample)
print("Prediction:", prediction)
if prediction == 0:
    print("Predicted Class: Setosa")
else:
    print("Predicted Class: Versicolor")
```
